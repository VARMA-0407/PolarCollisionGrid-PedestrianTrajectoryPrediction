# Measurement enhancement (appended cells)

Everything **above** this cell is the unchanged historical notebook (cells and saved outputs). Everything **below** was appended to add research-grade measurement instrumentation. Nothing below re-runs historical training, overwrites a checkpoint, configuration or prediction file, or modifies the PCG source.

**Safe by default.** Running these cells top to bottom only *reads* files and writes new files into a fresh directory (`MEAS_DIR`). Two optional cells can start real work, and both are OFF:

| Flag | Starts | Needs |
|---|---|---|
| `RUN_MONITORED_TRAINING` | a NEW isolated training run with resource monitoring (Part D) | the flag, the exact run id in `CONFIRM_RUN_ID`, GPU runtime, ~13 min for 5 epochs |
| `RUN_INFERENCE_BENCHMARK` | single-sample predictor timing of the epoch-4 checkpoint (Part E) | the flag, GPU runtime |

**Interpreter warning.** The historical notebook used two interpreters: the Colab default (Python 3.13.16, PyTorch 2.14.1+cu130) and `/content/pcg-env` (Python 3.11, PyTorch 2.2.1+cu121, the version the README asks for). Two complete 5-epoch trainings exist in the saved outputs (the cell with `/content/pcg-env/bin/python` and the later cell with plain `!python`); the archived checkpoints correspond to the **later** one. Every new run records the exact interpreter it actually used. Set `PYTHON_FOR_NEW_RUNS` explicitly.

**Labels used throughout.** *Retrospective* = extracted from outputs saved in this notebook (no new execution). *New run* = collected by the instruments below. Historical test metrics use `sample_size=5` with best-of-five **oracle** selection (ground truth picks the best sample) and are not single-sample results; the baseline test path also runs with dropout active (`net.eval()` is never called) and uses future actor information. Trajectory-only timing is **not** video or real-time performance.


```python
# ==============================================================================================
# MEASUREMENT ENHANCEMENT - SETUP (appended; nothing above this cell was modified)
# No training, no test.py run and no artifact write happens unless a flag below is set to True.
# ==============================================================================================
import os, sys, json, time, pickle, platform, subprocess, shutil, hashlib, datetime, pathlib

# ---- switches (all OFF) -------------------------------------------------------------------------
RUN_MONITORED_TRAINING = False   # Part D: starts a NEW isolated training run (GPU, minutes). Leave False unless intended.
CONFIRM_RUN_ID = ""              # to start training ALSO set this to the exact MEAS_RUN_ID printed below
RUN_INFERENCE_BENCHMARK = False  # Part E: inference-only timing of the epoch-4 checkpoint (isolated, read-only on artifacts)
TRAIN_EPOCHS_FOR_NEW_RUN = 5
PYTHON_FOR_NEW_RUNS = None       # None -> this kernel's interpreter. For comparability with the README use "/content/pcg-env/bin/python"
BENCH_WARMUP = 20                # warm-up trajectories excluded from timing
BENCH_MAX_TRAJECTORIES = None    # None -> all 1081 test sequences
BENCH_SEED = 12345

# ---- locations used by the historical notebook (read only here) ---------------------------------
REPO = "/content/PolarCollisionGrid-PedestrianTrajectoryPrediction"
DRIVE_BACKUP = "/content/drive/MyDrive/PCG_Experiments/PCG_T4_5_Epoch_Run"
STORE = os.path.join(REPO, "Store_Results")

# ---- fresh output directory for THIS measurement session -----------------------------------------
MEAS_RUN_ID = "meas_" + datetime.datetime.now(datetime.timezone.utc).strftime("%Y%m%dT%H%M%SZ")
OUT_PARENT = "/content/drive/MyDrive/PCG_Experiments/measurements" if os.path.isdir("/content/drive/MyDrive") else "/content/pcg_measurements"
MEAS_DIR = os.path.join(OUT_PARENT, MEAS_RUN_ID)
WORK_ROOT = os.path.join("/content/pcg_runs", MEAS_RUN_ID)       # local disk (fast) for any new run's working directories
os.makedirs(OUT_PARENT, exist_ok=True)
os.makedirs(MEAS_DIR, exist_ok=False)                             # fails if it exists: never reuse or overwrite
TOOLS_DIR = os.path.join(MEAS_DIR, "tools")
os.makedirs(TOOLS_DIR, exist_ok=False)

# ---- helper sources (embedded so this notebook is self-contained; written exclusively, never over an existing file)
HELPER_SOURCES = {
    "measurement_lib.py": r'''"""Shared, dependency-light helpers for the PCG measurement notebook (stdlib only; numpy optional).

Nothing here trains a model or touches an experiment artifact. Everything that writes uses exclusive creation
(`make_fresh_dir`, `write_new_json`) so an existing file or directory is never overwritten.
"""

import csv
import hashlib
import json
import math
import os
import re
import time
from datetime import datetime, timezone
from typing import Any, Dict, Iterable, List, Mapping, Optional, Sequence, Tuple

LIB_VERSION = "pcg-measurement-lib-1.0"


# ------------------------------------------------------------------ hashing / file records
def sha256_file(path: str, chunk: int = 1 << 20) -> str:
    h = hashlib.sha256()
    with open(path, "rb") as fh:
        for block in iter(lambda: fh.read(chunk), b""):
            h.update(block)
    return h.hexdigest()


def utc_now_iso() -> str:
    return datetime.now(timezone.utc).strftime("%Y-%m-%dT%H:%M:%SZ")


def file_record(path: str, label: str = "", expected_sha256: Optional[str] = None, expected_source: Optional[str] = None) -> Dict[str, Any]:
    """Read-only description of one file. ``matches_previous_hash`` is None when no previous hash is known."""
    rec: Dict[str, Any] = {"label": label, "path": path, "exists": os.path.isfile(path)}
    try:
        rec["resolved_path"] = os.path.realpath(path)
    except OSError:
        rec["resolved_path"] = None
    if rec["exists"]:
        st = os.stat(path)
        rec["size_bytes"] = st.st_size
        rec["modified_utc"] = datetime.fromtimestamp(st.st_mtime, timezone.utc).strftime("%Y-%m-%dT%H:%M:%SZ")
        rec["sha256"] = sha256_file(path)
    else:
        rec["size_bytes"] = rec["modified_utc"] = rec["sha256"] = None
    rec["previous_sha256"] = expected_sha256
    rec["previous_hash_source"] = expected_source if expected_sha256 else None
    rec["matches_previous_hash"] = None if (expected_sha256 is None or not rec["exists"]) else (rec["sha256"].lower() == expected_sha256.lower())
    return rec


# ----------------------------------------------------------------------- safe writing
def make_fresh_dir(path: str) -> str:
    """Create ``path`` (and parents). Refuses an existing non-empty directory; an existing EMPTY one is accepted."""
    if os.path.exists(path):
        if not os.path.isdir(path):
            raise FileExistsError("output path exists and is not a directory: %s" % path)
        if os.listdir(path):
            raise FileExistsError("output directory is not empty (nothing is overwritten or deleted): %s" % path)
    os.makedirs(path, exist_ok=True)
    return path


def write_new_json(path: str, obj: Any) -> str:
    """Exclusive create ('x'): raises FileExistsError instead of replacing a file. Returns the SHA-256."""
    text = json.dumps(obj, indent=2, sort_keys=True, allow_nan=False, default=str) + "\n"
    with open(path, "x", encoding="utf-8", newline="\n") as fh:
        fh.write(text)
    return hashlib.sha256(text.encode("utf-8")).hexdigest()


def write_new_csv(path: str, rows: Sequence[Mapping[str, Any]], fieldnames: Sequence[str]) -> None:
    with open(path, "x", newline="", encoding="utf-8") as fh:
        w = csv.DictWriter(fh, fieldnames=list(fieldnames))
        w.writeheader()
        for r in rows:
            w.writerow({k: r.get(k) for k in fieldnames})


# ------------------------------------------------------------------------- statistics
def percentile_nearest_rank(values: Sequence[float], q: float) -> float:
    if not values:
        raise ValueError("percentile of an empty sequence")
    if not 0.0 <= q <= 100.0:
        raise ValueError("q must be in [0, 100]")
    s = sorted(values)
    rank = max(1, int(math.ceil(q / 100.0 * len(s))))
    return s[min(rank, len(s)) - 1]


def describe(values: Sequence[float]) -> Dict[str, Any]:
    """count, mean, std (population), min, p50, p90, p95, p99, max. Percentiles are NEAREST-RANK (no interpolation)."""
    v = [float(x) for x in values if x is not None and math.isfinite(float(x))]
    if not v:
        return {"count": 0}
    n = len(v)
    mean = sum(v) / n
    std = math.sqrt(sum((x - mean) ** 2 for x in v) / n)
    return {"count": n, "mean": mean, "std": std, "min": min(v), "p50": percentile_nearest_rank(v, 50), "p90": percentile_nearest_rank(v, 90),
            "p95": percentile_nearest_rank(v, 95), "p99": percentile_nearest_rank(v, 99), "max": max(v), "percentile_method": "nearest_rank"}


# --------------------------------------------------------------------------- log parsing
_TRAIN_RE = re.compile(r"^\s*(\d+)/(\d+)\s+\(epoch\s+(\d+)\),\s*train_loss\s*=\s*(-?[\d.]+),\s*time/batch\s*=\s*([\d.]+)\s*$")
_GRID_RE = re.compile(r"grid calculation time for all the data:\s*([\d.]+)\s*seconds")
_TOTAL_RE = re.compile(r"The whole trainig time for\s+(\d+)\s+iteraction was\s+([\d.]+)\s+seconds")
_TEST_RE = re.compile(r"Processed trajectory number\s*:\s*(\d+)\s+out of\s+(\d+)\s+trajectories in time\s+([\d.]+)")


def parse_train_log(text: str) -> Dict[str, Any]:
    """Parse train.py stdout. Only values that are literally printed are returned; nothing is inferred."""
    batches = []
    for line in text.splitlines():
        m = _TRAIN_RE.match(line)
        if m:
            batches.append({"global_batch": int(m.group(1)), "total_batches": int(m.group(2)), "epoch": int(m.group(3)),
                            "train_loss": float(m.group(4)), "time_per_batch_s": float(m.group(5))})
    g, t = _GRID_RE.search(text), _TOTAL_RE.search(text)
    return {"batches": batches, "grid_calculation_s": float(g.group(1)) if g else None,
            "total_training_s": float(t.group(2)) if t else None, "epochs_reported": int(t.group(1)) if t else None}


def parse_test_log(text: str) -> Dict[str, Any]:
    rows = [{"trajectory": int(m.group(1)), "of": int(m.group(2)), "time_s": float(m.group(3))} for m in _TEST_RE.finditer(text)]
    return {"trajectories": rows}


def per_epoch_from_batches(batches: Sequence[Mapping[str, Any]]) -> List[Dict[str, Any]]:
    by: Dict[int, List[Mapping[str, Any]]] = {}
    for b in batches:
        by.setdefault(int(b["epoch"]), []).append(b)
    out = []
    for e in sorted(by):
        bs = by[e]
        t = [x["time_per_batch_s"] for x in bs]
        l = [x["train_loss"] for x in bs]
        out.append({"epoch": e, "n_batches": len(bs), "mean_loss": sum(l) / len(l), "sum_batch_time_s": sum(t), "batch_time": describe(t)})
    return out
''',
    "monitored_train_runner.py": r'''"""Monitored, ISOLATED training runner for the PCG baseline (notebook companion; train.py is NOT modified).

What it does: runs the unmodified repository ``train.py`` in-process inside a FRESH working directory (its own copy of ``Data/``
and its own ``Store_Results/``), while recording per-batch, per-epoch and resource measurements. It never runs by itself:
the notebook only calls it when the user sets an explicit flag and types the run id.

Isolation (train.py writes ``Store_Results/...`` and reads ``Data/...`` relative to the current directory, so a fresh
directory is what keeps existing checkpoints, configs, logs and results safe):
* the work directory must not exist or must be empty, must not be inside the repository, and gets its own ``Data/`` copy;
* no path of the original run is used; the repository's ``Store_Results`` tree is snapshotted before and verified after;
* a pre-flight returns a list of problems; any problem means the run is NOT started.

Measurement definitions (also written into the summary):
* batch time (sync) = CUDA-synchronised wall time from the start of a batch (``Module.zero_grad`` call) to the end of
  ``optimizer.step`` of the same batch. Data slicing before zero_grad, plotting, printing and checkpoint saving are outside it.
* epoch wall time = time between consecutive "epoch beginning" lines (the last epoch ends at the final training-time line); it
  includes batch creation, plotting and the checkpoint save.
* data loading/preprocessing = timed ``DataLoader.__init__`` (filtering), ``grid_creation`` and each epoch's ``batch_creater``.
* GPU utilisation / used memory = sampled from NVML or ``nvidia-smi`` every ``interval_s`` seconds (device-wide, not per process);
  allocated / reserved / peak memory = PyTorch allocator counters of THIS process.
* peak RAM = max process RSS among samples AND (Linux) the kernel's ``ru_maxrss``.
"""

import argparse
import csv
import hashlib
import importlib
import json
import math
import os
import platform
import shutil
import subprocess
import sys
import threading
import time
import traceback
from datetime import datetime, timezone
from typing import Any, Dict, List, Optional, Sequence

RUNNER_VERSION = "pcg-monitored-train-runner-1.0"
DEFAULT_REQUIRED = ("train.py", "utils/DataLoader.py", "Data/train/trajectories_train.cpkl")
DATA_FILES = ("Data/HBS/hbs.csv", "Data/train/trajectories.cpkl", "Data/train/trajectories_train.cpkl", "Data/test/trajectories_test.cpkl")
SOURCE_FILES = ("train.py", "test.py", "utils/DataLoader.py", "utils/helper.py", "utils/Interaction.py", "utils/grid.py", "visualization.py")


def _sha(path: str) -> str:
    h = hashlib.sha256()
    with open(path, "rb") as fh:
        for b in iter(lambda: fh.read(1 << 20), b""):
            h.update(b)
    return h.hexdigest()


def _utc() -> str:
    return datetime.now(timezone.utc).strftime("%Y-%m-%dT%H:%M:%SZ")


def _pct(values: Sequence[float], q: float) -> float:
    s = sorted(values)
    return s[min(len(s), max(1, int(math.ceil(q / 100.0 * len(s))))) - 1]


def stats(values: Sequence[float]) -> Dict[str, Any]:
    v = [float(x) for x in values if x is not None and math.isfinite(float(x))]
    if not v:
        return {"count": 0}
    m = sum(v) / len(v)
    return {"count": len(v), "mean": m, "std": math.sqrt(sum((x - m) ** 2 for x in v) / len(v)), "min": min(v), "p50": _pct(v, 50),
            "p90": _pct(v, 90), "p95": _pct(v, 95), "max": max(v), "percentile_method": "nearest_rank"}


def tree_snapshot(root: str) -> Dict[str, List[Any]]:
    """path -> [size, mtime_ns] for every file under root (used to prove the original run directory was not touched)."""
    snap: Dict[str, List[Any]] = {}
    if os.path.isdir(root):
        for d, _, fs in os.walk(root):
            for f in fs:
                p = os.path.join(d, f)
                try:
                    st = os.stat(p)
                    snap[os.path.relpath(p, root)] = [st.st_size, st.st_mtime_ns]
                except OSError:
                    snap[os.path.relpath(p, root)] = None
    return snap


def preflight(repo: str, workdir: str, required_files: Sequence[str] = DEFAULT_REQUIRED) -> List[str]:
    """Return a list of problems. An empty list means isolation can be guaranteed; otherwise the run must not start."""
    problems: List[str] = []
    repo_r, work_r = os.path.realpath(repo), os.path.realpath(workdir)
    for rel in required_files:
        if not os.path.isfile(os.path.join(repo, rel)):
            problems.append("repository file missing: %s" % rel)
    if work_r == repo_r or work_r.startswith(repo_r + os.sep):
        problems.append("work directory must be OUTSIDE the repository (it is %s)" % work_r)
    if os.path.islink(workdir):
        problems.append("work directory is a symbolic link: %s" % workdir)
    if os.path.exists(workdir):
        if not os.path.isdir(workdir):
            problems.append("work directory path exists and is not a directory: %s" % workdir)
        elif os.listdir(workdir):
            problems.append("work directory is not empty (a checkpoint or log could be overwritten): %s" % workdir)
    anc = work_r
    while anc and not os.path.exists(anc) and os.path.dirname(anc) != anc:
        anc = os.path.dirname(anc)
    if os.path.exists(anc) and not os.path.isdir(anc):
        problems.append("an ancestor of the work directory is a file: %s" % anc)
    return problems


# --------------------------------------------------------------------------- GPU telemetry
class GpuProbe:
    """GPU utilisation and device memory. Tries NVML, then nvidia-smi. Reports its source honestly."""

    def __init__(self):
        self.source = "unavailable"
        self._nvml = None
        try:
            import pynvml  # type: ignore
            pynvml.nvmlInit()
            self._nvml = (pynvml, pynvml.nvmlDeviceGetHandleByIndex(0))
            self.source = "pynvml"
        except Exception:
            self._nvml = None
            try:
                subprocess.run(["nvidia-smi", "-L"], capture_output=True, timeout=5, check=True)
                self.source = "nvidia-smi"
            except Exception:
                self.source = "unavailable"

    def read(self):
        """(utilization_percent, memory_used_mib, memory_total_mib) or (None, None, None)."""
        try:
            if self._nvml is not None:
                nv, h = self._nvml
                u = nv.nvmlDeviceGetUtilizationRates(h)
                m = nv.nvmlDeviceGetMemoryInfo(h)
                return float(u.gpu), m.used / 2 ** 20, m.total / 2 ** 20
            if self.source == "nvidia-smi":
                out = subprocess.run(["nvidia-smi", "--query-gpu=utilization.gpu,memory.used,memory.total", "--format=csv,noheader,nounits"],
                                     capture_output=True, text=True, timeout=5).stdout.strip().splitlines()[0]
                a, b, c = [float(x) for x in out.split(",")]
                return a, b, c
        except Exception:
            pass
        return None, None, None


def gpu_driver_version() -> Optional[str]:
    try:
        return subprocess.run(["nvidia-smi", "--query-gpu=driver_version", "--format=csv,noheader"], capture_output=True, text=True,
                              timeout=5).stdout.strip().splitlines()[0]
    except Exception:
        return None


# ---------------------------------------------------------------------------- the monitor
class Tee:
    def __init__(self, original, path, on_line):
        self._o, self._f, self._cb, self._buf = original, open(path, "x", encoding="utf-8"), on_line, ""

    def write(self, s):
        self._o.write(s)
        self._f.write(s)
        self._buf += s
        while "\n" in self._buf:
            line, self._buf = self._buf.split("\n", 1)
            self._cb(line)
        return len(s)

    def flush(self):
        self._o.flush()
        self._f.flush()

    def close(self):
        if self._buf:
            self._cb(self._buf)
        self._f.close()

    def __getattr__(self, name):
        return getattr(self._o, name)


class Monitor:
    def __init__(self, torch_mod, interval_s: float = 1.0, batch_size: int = 10):
        self.torch, self.interval_s, self.batch_size = torch_mod, interval_s, batch_size
        self.cuda = bool(torch_mod.cuda.is_available())
        self.t0 = time.perf_counter()
        self.samples: List[Dict[str, Any]] = []
        self.batches: List[Dict[str, Any]] = []
        self.phases: List[Dict[str, Any]] = []
        self.markers: List[Dict[str, Any]] = []
        self.reported: List[Dict[str, Any]] = []
        self.save_times: List[float] = []
        self._batch_t0: Optional[float] = None
        self._epoch = -1
        self._global_batch = -1
        self._stop = threading.Event()
        self._thread: Optional[threading.Thread] = None
        self.gpu = GpuProbe()
        try:
            import psutil  # type: ignore
            self.psutil, self.proc = psutil, psutil.Process(os.getpid())
            self.proc.cpu_percent(None)
            psutil.cpu_percent(None)
        except Exception:
            self.psutil = self.proc = None

    # -- timing primitives
    def _sync(self):
        if self.cuda:
            self.torch.cuda.synchronize()

    def now(self) -> float:
        return time.perf_counter() - self.t0

    def mark(self, kind: str, **kw):
        self.markers.append(dict(kind=kind, t_s=self.now(), **kw))

    # -- stdout parsing
    def on_line(self, line: str):
        import re
        m = re.match(r"^\s*(\d+)/(\d+)\s+\(epoch\s+(\d+)\),\s*train_loss\s*=\s*(-?[\d.]+),\s*time/batch\s*=\s*([\d.]+)\s*$", line)
        if m:
            self.reported.append({"global_batch": int(m.group(1)), "epoch": int(m.group(3)), "train_loss": float(m.group(4)),
                                  "reported_time_per_batch_s": float(m.group(5)), "t_s": self.now()})
            return
        if "Training epoch beginning" in line:
            self._epoch += 1
            self.mark("epoch_begin", epoch=self._epoch)
        elif line.strip() == "Saving model":
            self.mark("saving_model", epoch=self._epoch)
        elif "The whole trainig time" in line:
            self.mark("train_done")

    # -- hooks
    def patched_zero_grad(self, orig):
        mon = self

        def zero_grad(module, *a, **k):
            mon._sync()
            mon._batch_t0 = time.perf_counter()
            return orig(module, *a, **k)
        return zero_grad

    def patched_step(self, orig):
        mon = self

        def step(opt, *a, **k):
            r = orig(opt, *a, **k)
            mon._sync()
            t1 = time.perf_counter()
            if mon._batch_t0 is not None:
                mon._global_batch += 1
                mon.batches.append({"global_batch": mon._global_batch, "epoch": max(mon._epoch, 0), "sync_batch_time_s": t1 - mon._batch_t0,
                                    "t_end_s": t1 - mon.t0})
                mon._batch_t0 = None
            return r
        return step

    def timed(self, name: str, orig):
        mon = self

        def wrapper(*a, **k):
            mon._sync()
            t = time.perf_counter()
            try:
                return orig(*a, **k)
            finally:
                mon._sync()
                mon.phases.append({"phase": name, "epoch": max(mon._epoch, 0), "duration_s": time.perf_counter() - t, "t_start_s": t - mon.t0})
        return wrapper

    def timed_save(self, orig):
        mon = self

        def save(*a, **k):
            t = time.perf_counter()
            try:
                return orig(*a, **k)
            finally:
                mon.save_times.append(time.perf_counter() - t)
        return save

    # -- sampler thread
    def start_sampler(self):
        def loop():
            while not self._stop.is_set():
                self.samples.append(self.sample())
                self._stop.wait(self.interval_s)
        self._thread = threading.Thread(target=loop, daemon=True)
        self._thread.start()

    def stop_sampler(self):
        self._stop.set()
        if self._thread is not None:
            self._thread.join(timeout=10)
        self.samples.append(self.sample())

    def sample(self) -> Dict[str, Any]:
        u, used, total = self.gpu.read()
        mib = 2 ** 20
        row: Dict[str, Any] = {"t_s": round(self.now(), 3), "utc": _utc(), "epoch": max(self._epoch, 0), "global_batch": self._global_batch,
                               "gpu_util_pct": u, "gpu_mem_used_mib": used, "gpu_mem_total_mib": total}
        if self.cuda:
            row["torch_allocated_mib"] = self.torch.cuda.memory_allocated() / mib
            row["torch_reserved_mib"] = self.torch.cuda.memory_reserved() / mib
            row["torch_max_allocated_mib"] = self.torch.cuda.max_memory_allocated() / mib
        else:
            row["torch_allocated_mib"] = row["torch_reserved_mib"] = row["torch_max_allocated_mib"] = None
        if self.psutil is not None:
            row["cpu_pct_system"] = self.psutil.cpu_percent(None)
            row["cpu_pct_process"] = self.proc.cpu_percent(None)
            row["process_rss_mib"] = self.proc.memory_info().rss / mib
            vm = self.psutil.virtual_memory()
            row["ram_used_mib"], row["ram_total_mib"] = (vm.total - vm.available) / mib, vm.total / mib
        else:
            row["cpu_pct_system"] = row["cpu_pct_process"] = row["process_rss_mib"] = row["ram_used_mib"] = row["ram_total_mib"] = None
        return row


# ------------------------------------------------------------------------------- summary
def build_epoch_rows(mon: Monitor) -> List[Dict[str, Any]]:
    rows = []
    begins = [m for m in mon.markers if m["kind"] == "epoch_begin"]
    done = [m for m in mon.markers if m["kind"] == "train_done"]
    end_of_run = done[-1]["t_s"] if done else (mon.batches[-1]["t_end_s"] if mon.batches else None)
    by_epoch: Dict[int, List[Dict[str, Any]]] = {}
    for b in mon.batches:
        by_epoch.setdefault(b["epoch"], []).append(b)
    rep: Dict[int, List[Dict[str, Any]]] = {}
    for r in mon.reported:
        rep.setdefault(r["epoch"], []).append(r)
    for i, bg in enumerate(begins):
        e = bg["epoch"]
        end = begins[i + 1]["t_s"] if i + 1 < len(begins) else end_of_run
        bt = [b["sync_batch_time_s"] for b in by_epoch.get(e, [])]
        losses = [r["train_loss"] for r in rep.get(e, [])]
        wall = (end - bg["t_s"]) if end is not None else None
        compute = sum(bt)
        prep = sum(p["duration_s"] for p in mon.phases if p["phase"] == "batch_creater" and p["epoch"] == e)
        s = stats(bt)
        rows.append({"epoch": e, "n_batches": len(bt), "epoch_wall_s": wall, "sum_batch_time_s": compute, "batch_creater_s": prep,
                     "batch_time_p50_s": s.get("p50"), "batch_time_p90_s": s.get("p90"), "batch_time_p95_s": s.get("p95"), "batch_time_max_s": s.get("max"),
                     "mean_train_loss": (sum(losses) / len(losses)) if losses else None,
                     "batches_per_s_compute": (len(bt) / compute) if compute > 0 else None,
                     "sequences_per_s_compute": (len(bt) * mon.batch_size / compute) if compute > 0 else None,
                     "sequences_per_s_wall": (len(bt) * mon.batch_size / wall) if wall else None})
    return rows


def hash_tree(root: str, rels: Sequence[str]) -> Dict[str, Optional[str]]:
    return {r: (_sha(os.path.join(root, r)) if os.path.isfile(os.path.join(root, r)) else None) for r in rels}


def _write_csv(path: str, rows: Sequence[Dict[str, Any]], fields: Sequence[str]) -> None:
    with open(path, "x", newline="", encoding="utf-8") as fh:
        w = csv.DictWriter(fh, fieldnames=list(fields))
        w.writeheader()
        for r in rows:
            w.writerow({k: r.get(k) for k in fields})


# --------------------------------------------------------------------------------- runner
def run_monitored(repo: str, workdir: str, run_id: str, num_epochs: int, *, method: int = 4, batch_size: int = 10, interval_s: float = 1.0,
                  extra_args: Sequence[str] = (), train_module: str = "train", required_files: Sequence[str] = DEFAULT_REQUIRED,
                  data_dir: str = "Data") -> Dict[str, Any]:
    """Run train.py monitored in a fresh work directory. Returns the summary dict (also written to disk)."""
    problems = preflight(repo, workdir, required_files)
    if problems:
        raise RuntimeError("isolation cannot be guaranteed; the run was NOT started:\n  - " + "\n  - ".join(problems))
    repo = os.path.realpath(repo)
    workdir = os.path.realpath(workdir)
    os.makedirs(workdir, exist_ok=True)               # creates missing parents; the directory itself is new or empty (checked above)
    out = os.path.join(workdir, "measurements")
    os.makedirs(out)                                              # fresh; fails if it somehow exists
    for sub in ("Store_Results/log", "Store_Results/model", "Store_Results/plot/train"):
        os.makedirs(os.path.join(workdir, sub))
    if os.path.isdir(os.path.join(repo, data_dir)):
        shutil.copytree(os.path.join(repo, data_dir), os.path.join(workdir, data_dir))     # isolated copy: train.py may write next to its data
    repo_results_before = tree_snapshot(os.path.join(repo, "Store_Results"))
    src_hash_before = hash_tree(repo, SOURCE_FILES)
    data_hash_before = hash_tree(repo, DATA_FILES)

    summary: Dict[str, Any] = {"runner_version": RUNNER_VERSION, "run_id": run_id, "started_utc": _utc(), "repo": repo, "workdir": workdir,
                               "command": {"module": train_module, "method": method, "num_epochs": num_epochs, "extra_args": list(extra_args)},
                               "status": "running"}
    old_cwd, old_argv, old_out = os.getcwd(), list(sys.argv), sys.stdout
    mon = None
    tee = None
    err = None
    try:
        os.chdir(workdir)
        sys.path.insert(0, repo)
        import torch  # noqa: WPS433
        mon = Monitor(torch, interval_s, batch_size)
        summary["environment"] = {
            "python_executable": sys.executable, "python_version": sys.version, "platform": platform.platform(), "torch": torch.__version__,
            "torch_cuda": torch.version.cuda, "cudnn": (torch.backends.cudnn.version() if torch.cuda.is_available() else None),
            "cuda_available": mon.cuda, "gpu_name": torch.cuda.get_device_name(0) if mon.cuda else None,
            "gpu_total_vram_mib": (torch.cuda.get_device_properties(0).total_memory / 2 ** 20) if mon.cuda else None,
            "gpu_driver": gpu_driver_version(), "gpu_telemetry_source": mon.gpu.source,
            "seeds": "none set by train.py or this runner (the baseline code sets no seed); runs are not bitwise reproducible",
            "cudnn_deterministic": torch.backends.cudnn.deterministic, "cudnn_benchmark": torch.backends.cudnn.benchmark}
        try:
            import numpy, pandas  # noqa: F401
            summary["environment"].update({"numpy": numpy.__version__, "pandas": pandas.__version__})
        except Exception:
            pass
        if mon.cuda:
            torch.cuda.reset_peak_memory_stats()
        # --- instrumentation (class-level patches; train.py itself is untouched)
        orig_zero_grad = torch.nn.Module.zero_grad
        orig_step = torch.optim.RMSprop.step
        orig_save = torch.save
        torch.nn.Module.zero_grad = mon.patched_zero_grad(orig_zero_grad)
        torch.optim.RMSprop.step = mon.patched_step(orig_step)
        torch.save = mon.timed_save(orig_save)
        dl_mod = importlib.import_module("utils.DataLoader")
        dl_cls = dl_mod.DataLoader
        patched = {}
        for name in ("__init__", "grid_creation", "batch_creater"):
            if hasattr(dl_cls, name):
                patched[name] = getattr(dl_cls, name)
                setattr(dl_cls, name, mon.timed("DataLoader." + name if name != "batch_creater" else "batch_creater", patched[name]))
        tmod = importlib.import_module(train_module)
        tee = Tee(old_out, os.path.join(out, "training_stdout.log"), mon.on_line)
        sys.stdout = tee
        sys.argv = [train_module + ".py", "--method", str(method), "--num_epochs", str(num_epochs), *list(extra_args)]    # no other option is added
        mon.start_sampler()
        t_wall0 = time.perf_counter()
        try:
            tmod.main()
        finally:
            mon._sync()
            summary["train_main_wall_s"] = time.perf_counter() - t_wall0
            mon.stop_sampler()
            sys.stdout = old_out
            tee.close()
            torch.nn.Module.zero_grad = orig_zero_grad
            torch.optim.RMSprop.step = orig_step
            torch.save = orig_save
            for name, fn in patched.items():
                setattr(dl_cls, name, fn)
        summary["status"] = "completed"
    except BaseException as exc:                                  # noqa: BLE001 - recorded, then re-raised after writing outputs
        err = exc
        summary["status"] = "failed"
        summary["error"] = {"type": type(exc).__name__, "message": str(exc), "traceback": traceback.format_exc()}
    finally:
        sys.stdout = old_out
        sys.argv = old_argv
        os.chdir(old_cwd)
        if sys.path and sys.path[0] == repo:
            sys.path.pop(0)

    summary["finished_utc"] = _utc()
    cfg_path = os.path.join(workdir, "Store_Results", "model", "config.pkl")
    if os.path.isfile(cfg_path):
        try:
            import pickle
            with open(cfg_path, "rb") as fh:
                ns = pickle.load(fh)                              # the Namespace train.py itself wrote in the fresh work directory
            summary["training_args"] = {k: (v if isinstance(v, (int, float, str, bool, list, type(None))) else str(v)) for k, v in vars(ns).items()}
            if mon is not None and isinstance(getattr(ns, "batch_size", None), int):
                mon.batch_size = ns.batch_size                    # throughput uses the batch size train.py actually used
        except Exception as exc:                                  # noqa: BLE001
            summary["training_args_error"] = str(exc)
    if mon is not None:
        epoch_rows = build_epoch_rows(mon)
        sync_t = [b["sync_batch_time_s"] for b in mon.batches]
        rep_t = [r["reported_time_per_batch_s"] for r in mon.reported]
        by_gb = {r["global_batch"]: r for r in mon.reported}
        batch_rows = [{"global_batch": b["global_batch"], "epoch": b["epoch"], "sync_batch_time_s": b["sync_batch_time_s"],
                       "reported_time_per_batch_s": by_gb.get(b["global_batch"], {}).get("reported_time_per_batch_s"),
                       "train_loss": by_gb.get(b["global_batch"], {}).get("train_loss")} for b in mon.batches]
        res_fields = ["t_s", "utc", "epoch", "global_batch", "gpu_util_pct", "gpu_mem_used_mib", "gpu_mem_total_mib", "torch_allocated_mib", "torch_reserved_mib",
                      "torch_max_allocated_mib", "cpu_pct_system", "cpu_pct_process", "process_rss_mib", "ram_used_mib", "ram_total_mib"]
        _write_csv(os.path.join(out, "training_resource_log.csv"), mon.samples, res_fields)
        _write_csv(os.path.join(out, "training_batch_log.csv"), batch_rows, ["global_batch", "epoch", "sync_batch_time_s", "reported_time_per_batch_s", "train_loss"])
        _write_csv(os.path.join(out, "training_epoch_summary.csv"), epoch_rows,
                   ["epoch", "n_batches", "epoch_wall_s", "sum_batch_time_s", "batch_creater_s", "batch_time_p50_s", "batch_time_p90_s", "batch_time_p95_s",
                    "batch_time_max_s", "mean_train_loss", "batches_per_s_compute", "sequences_per_s_compute", "sequences_per_s_wall"])
        col = lambda k: [s[k] for s in mon.samples if s.get(k) is not None]          # noqa: E731
        peak_rss_kernel = None
        try:
            import resource
            peak_rss_kernel = resource.getrusage(resource.RUSAGE_SELF).ru_maxrss / 1024.0
        except Exception:
            pass
        total_batch = sum(sync_t)
        summary.update({
            "batches_recorded": len(sync_t), "epochs_recorded": len(epoch_rows),
            "batch_time_sync_s": stats(sync_t), "batch_time_sync_excluding_first_s": stats(sync_t[1:]),
            "batch_time_reported_by_train_py_s": stats(rep_t),
            "throughput": {"sequences_per_s_compute": (len(sync_t) * mon.batch_size / total_batch) if total_batch > 0 else None,
                           "batches_per_s_compute": (len(sync_t) / total_batch) if total_batch > 0 else None,
                           "definition": "batches*batch_size / sum of synchronised batch times"},
            "data_preprocessing_s": {p: sum(x["duration_s"] for x in mon.phases if x["phase"] == p) for p in sorted({x["phase"] for x in mon.phases})},
            "checkpoint_save_s": {"count": len(mon.save_times), "total": sum(mon.save_times)},
            "gpu_utilisation_pct": stats(col("gpu_util_pct")), "gpu_device_memory_used_mib": stats(col("gpu_mem_used_mib")),
            "torch_allocated_mib": stats(col("torch_allocated_mib")), "torch_reserved_mib": stats(col("torch_reserved_mib")),
            "torch_peak_allocated_mib": (max(col("torch_max_allocated_mib")) if col("torch_max_allocated_mib") else None),
            "cpu_pct_system": stats(col("cpu_pct_system")), "cpu_pct_process": stats(col("cpu_pct_process")),
            "peak_process_rss_mib_sampled": (max(col("process_rss_mib")) if col("process_rss_mib") else None),
            "peak_process_rss_mib_kernel_ru_maxrss": peak_rss_kernel,
            "peak_system_ram_used_mib_sampled": (max(col("ram_used_mib")) if col("ram_used_mib") else None),
            "sampler": {"interval_s": interval_s, "n_samples": len(mon.samples), "gpu_source": mon.gpu.source},
            "markers": mon.markers,
            "caveats": ["GPU utilisation and used memory are device-wide samples, not per-process counters",
                        "sampling can miss short peaks; torch peak counters and ru_maxrss are exact for their scope",
                        "the sampler thread and CUDA synchronisation at batch boundaries add small overhead",
                        "this is baseline PCG TRAINING cost on the declared hardware; it says nothing about video or deployment performance"]})
        if tee is not None:
            summary["stdout_log"] = "training_stdout.log"
    # --- isolation verification and hashes
    repo_results_after = tree_snapshot(os.path.join(repo, "Store_Results"))
    summary["isolation_check"] = {
        "repo_store_results_unchanged": repo_results_before == repo_results_after,
        "repo_source_hashes_unchanged": src_hash_before == hash_tree(repo, SOURCE_FILES),
        "repo_data_hashes_unchanged": data_hash_before == hash_tree(repo, DATA_FILES),
        "workdir_data_hashes_equal_repo_data": data_hash_before == hash_tree(workdir, DATA_FILES)}
    summary["source_sha256"] = src_hash_before
    summary["data_sha256"] = data_hash_before
    produced = {}
    for d, _, fs in os.walk(os.path.join(workdir, "Store_Results")):
        for f in fs:
            p = os.path.join(d, f)
            produced[os.path.relpath(p, workdir)] = {"size_bytes": os.path.getsize(p), "sha256": _sha(p)}
    summary["produced_artifacts"] = produced
    with open(os.path.join(out, "training_resource_summary.json"), "x", encoding="utf-8") as fh:
        json.dump(summary, fh, indent=2, sort_keys=True, default=str)
    if err is not None:
        raise err
    return summary


def main(argv: Optional[Sequence[str]] = None) -> int:
    ap = argparse.ArgumentParser(description="Monitored isolated PCG training run (never run automatically).")
    ap.add_argument("--repo", required=True)
    ap.add_argument("--workdir", required=True, help="fresh or empty directory OUTSIDE the repository")
    ap.add_argument("--run-id", required=True)
    ap.add_argument("--num-epochs", type=int, required=True)
    ap.add_argument("--method", type=int, default=4)
    ap.add_argument("--batch-size", type=int, default=10)
    ap.add_argument("--interval-s", type=float, default=1.0)
    ap.add_argument("--preflight-only", action="store_true")
    a, extra = ap.parse_known_args(argv)
    if a.preflight_only:
        probs = preflight(a.repo, a.workdir)
        print(json.dumps({"problems": probs, "safe_to_run": not probs}, indent=2))
        return 0 if not probs else 2
    try:
        s = run_monitored(a.repo, a.workdir, a.run_id, a.num_epochs, method=a.method, batch_size=a.batch_size, interval_s=a.interval_s, extra_args=extra)
    except RuntimeError as exc:
        print(str(exc), file=sys.stderr)
        return 2
    print(json.dumps({"status": s["status"], "workdir": a.workdir}, indent=2))
    return 0


if __name__ == "__main__":
    sys.exit(main())
''',
    "inference_benchmark.py": r'''"""Single-sample PCG trajectory-predictor inference benchmark (notebook companion; test.py is NOT modified or run).

SCOPE (written into every output): this measures the PCG trajectory-PREDICTOR code path only, one prediction sample per input
(``sample_size = 1``). It is NOT end-to-end video performance (no detection, tracking or video decoding) and it is NOT a
deployment-valid latency: the baseline ``sample`` function and the preparation steps still consume the future actor information
that the historical test path uses (see the Phase 2 Task 4 audit). The numbers characterise this code path on the declared hardware.

Relation to ``test.py`` (read at the commit recorded in the output): the per-trajectory steps are REPLICATED here, step for step,
from ``test.py`` ("dense vector creation" ... "revert position change"), and the baseline ``sample`` function is imported from
``test.py`` unchanged. Left out on purpose: the ground-truth scoring (ADE/FDE/MHD/speed/heading/collision), the best-of-N
selection (best-of-N needs ground truth: that is the oracle) and the result list. Because ``test.py`` never calls ``net.eval()``,
this benchmark does not either unless ``eval_mode=True`` is requested; the model contains ``nn.Dropout`` so the baseline
prediction path runs with dropout ACTIVE. Both facts are recorded.

Gradient tracking (v1.1): the baseline ``sample`` function already runs EVERYTHING after its setup lines, including every ``net(...)``
call and its return, under ``torch.no_grad()`` (checked statically from test.py's AST by ``sample_grad_scope`` and observed at run
time). It does not use ``torch.inference_mode()``. The benchmark adds NO extra ``no_grad`` / ``inference_mode`` context and never
changes the model's train/eval state, so what is timed is exactly the baseline execution. The stages outside ``sample`` (preparation and
post-processing, as in test.py) run with grad mode enabled but create no autograd graph because no tensor in them requires grad. Every
output records the observed grad mode in ``gradient_tracking`` and the model's train/eval state before and after the run.

Timing method: ``torch.cuda.synchronize()`` before and after every timed stage when CUDA is used; ``time.perf_counter``.
Two separate passes over the same trajectories with the same seed:
  pass 1 "pipeline": stages prep (CPU: dense arrays + interaction grids + position change), transfer (to GPU), rollout (the baseline
                     ``sample`` call: model forwards + Gaussian sampling + per-step grid work), post (revert position change);
                     pipeline latency = the whole per-trajectory span;
  pass 2 "model_only": forward hooks on the network time every ``net(...)`` call (synchronised) and sum them per trajectory.
Warm-up trajectories are processed first in each pass and excluded from all statistics.
"""

import argparse
import csv
import hashlib
import importlib.util
import json
import math
import os
import pickle
import platform
import random
import shutil
import subprocess
import sys
import time
from datetime import datetime, timezone
from typing import Any, Dict, List, Optional, Sequence

BENCH_VERSION = "pcg-inference-benchmark-1.1"
SCOPE = ("PCG trajectory-predictor-only, single-sample (sample_size=1), baseline code path. NOT end-to-end video performance. NOT deployment-valid: "
         "the baseline path still uses future actor information as in the historical test. Ground-truth scoring and oracle selection are excluded. "
         "Gradient tracking: see the gradient_tracking record (baseline sample() runs under torch.no_grad(); the benchmark adds no other context).")


def _sha(path: str) -> str:
    h = hashlib.sha256()
    with open(path, "rb") as fh:
        for b in iter(lambda: fh.read(1 << 20), b""):
            h.update(b)
    return h.hexdigest()


def _utc() -> str:
    return datetime.now(timezone.utc).strftime("%Y-%m-%dT%H:%M:%SZ")


def _pct(v: Sequence[float], q: float) -> float:
    s = sorted(v)
    return s[min(len(s), max(1, int(math.ceil(q / 100.0 * len(s))))) - 1]


def describe(values: Sequence[float]) -> Dict[str, Any]:
    v = [float(x) for x in values if x is not None and math.isfinite(float(x))]
    if not v:
        return {"count": 0}
    m = sum(v) / len(v)
    return {"count": len(v), "mean_s": m, "std_s": math.sqrt(sum((x - m) ** 2 for x in v) / len(v)), "min_s": min(v), "p50_s": _pct(v, 50), "p90_s": _pct(v, 90),
            "p95_s": _pct(v, 95), "p99_s": _pct(v, 99), "max_s": max(v), "percentile_method": "nearest_rank"}


def preflight(repo: str, workdir: str, checkpoint: str, config: str, required: Sequence[str] = ("test.py", "utils/DataLoader.py", "Data/test/trajectories_test.cpkl")) -> List[str]:
    problems = []
    for rel in required:
        if not os.path.isfile(os.path.join(repo, rel)):
            problems.append("repository file missing: %s" % rel)
    for label, p in (("checkpoint", checkpoint), ("config", config)):
        if not os.path.isfile(p):
            problems.append("%s file missing: %s" % (label, p))
    rr, wr = os.path.realpath(repo), os.path.realpath(workdir)
    if wr == rr or wr.startswith(rr + os.sep):
        problems.append("work directory must be OUTSIDE the repository: %s" % wr)
    if os.path.exists(workdir) and (not os.path.isdir(workdir) or os.listdir(workdir)):
        problems.append("work directory exists and is not an empty directory: %s" % workdir)
    return problems


_NO_GRAD_CONTEXTS = ("torch.no_grad()", "torch.set_grad_enabled(False)")


def sample_grad_scope(test_py_path: str) -> Dict[str, Any]:
    """Static (AST) check of the baseline ``sample`` function in test.py: which model calls run under a no-grad guard.

    Reads the file only. Returns a plain dict that is stored in every benchmark summary."""
    import ast
    tree = ast.parse(open(test_py_path, encoding="utf-8").read())
    fn = next((n for n in tree.body if isinstance(n, ast.FunctionDef) and n.name == "sample"), None)
    if fn is None:
        return {"function_found": False, "verdict": "sample() was not found in test.py"}
    guards = []
    for w in ast.walk(fn):
        if isinstance(w, ast.With):
            for item in w.items:
                ctx = ast.unparse(item.context_expr)
                if ctx in _NO_GRAD_CONTEXTS or ctx.startswith("torch.inference_mode("):
                    guards.append({"context": ctx, "lines": [w.lineno, w.end_lineno]})
    calls = [n for n in ast.walk(fn) if isinstance(n, ast.Call) and isinstance(n.func, ast.Name) and n.func.id == "net"]
    returns = [n for n in ast.walk(fn) if isinstance(n, ast.Return)]

    def inside(node):
        return any(g["lines"][0] <= node.lineno <= g["lines"][1] for g in guards)
    n_in = sum(1 for c in calls if inside(c))
    out = {"function_found": True, "function_lines": [fn.lineno, fn.end_lineno], "guards": guards, "net_calls_total": len(calls), "net_calls_inside_guard": n_in,
           "returns_total": len(returns), "returns_inside_guard": sum(1 for r in returns if inside(r)),
           "uses_inference_mode": any(g["context"].startswith("torch.inference_mode(") for g in guards), "uses_no_grad": any(g["context"] in _NO_GRAD_CONTEXTS for g in guards)}
    if guards and calls and n_in == len(calls):
        out["verdict"] = "all %d net(...) calls in sample() run under %s" % (len(calls), "/".join(sorted({g["context"] for g in guards})))
    else:
        out["verdict"] = "NOT all model calls in sample() are under a no-grad guard (%d of %d)" % (n_in, len(calls))
    return out


def _gpu_driver() -> Optional[str]:
    try:
        return subprocess.run(["nvidia-smi", "--query-gpu=driver_version", "--format=csv,noheader"], capture_output=True, text=True,
                              timeout=5).stdout.strip().splitlines()[0]
    except Exception:
        return None


def run_benchmark(repo: str, workdir: str, run_id: str, checkpoint: str, config: str, *, epoch: int, method: int = 4, warmup: int = 20,
                  max_trajectories: Optional[int] = None, seed: int = 12345, use_cuda: Optional[bool] = None, eval_mode: bool = False,
                  expected_checkpoint_sha256: Optional[str] = None, expected_config_sha256: Optional[str] = None,
                  hardware_label: Optional[str] = None) -> Dict[str, Any]:
    problems = preflight(repo, workdir, checkpoint, config)
    if problems:
        raise RuntimeError("benchmark NOT started:\n  - " + "\n  - ".join(problems))
    repo, workdir = os.path.realpath(repo), os.path.realpath(workdir)
    os.makedirs(workdir, exist_ok=True)
    out = os.path.join(workdir, "measurements")
    os.makedirs(out)
    model_dir = os.path.join(workdir, "Store_Results", "model", "CollisionGrid" if method == 4 else ("SocialLSTM" if method == 1 else "VanillaLSTM"))
    os.makedirs(model_dir)
    ck_dst = os.path.join(model_dir, "SOCIALLSTM_lstm_model_%d.tar" % epoch)
    cf_dst = os.path.join(model_dir, "config.pkl")
    shutil.copy2(checkpoint, ck_dst)                                  # COPIES: the originals are only read
    shutil.copy2(config, cf_dst)
    shutil.copytree(os.path.join(repo, "Data"), os.path.join(workdir, "Data"))
    hashes = {"checkpoint_sha256": _sha(checkpoint), "checkpoint_copy_sha256": _sha(ck_dst), "config_sha256": _sha(config), "config_copy_sha256": _sha(cf_dst),
              "test_py_sha256": _sha(os.path.join(repo, "test.py")), "test_data_sha256": _sha(os.path.join(repo, "Data", "test", "trajectories_test.cpkl"))}
    if hashes["checkpoint_sha256"] != hashes["checkpoint_copy_sha256"] or hashes["config_sha256"] != hashes["config_copy_sha256"]:
        raise RuntimeError("copied checkpoint/config hashes differ from the originals; aborting")
    hashes["checkpoint_matches_expected"] = None if not expected_checkpoint_sha256 else hashes["checkpoint_sha256"].lower() == expected_checkpoint_sha256.lower()
    hashes["config_matches_expected"] = None if not expected_config_sha256 else hashes["config_sha256"].lower() == expected_config_sha256.lower()

    old_cwd = os.getcwd()
    sys.path.insert(0, repo)
    try:
        os.chdir(workdir)
        import numpy as np
        import torch
        cuda = torch.cuda.is_available() if use_cuda is None else bool(use_cuda)
        if cuda and not torch.cuda.is_available():
            raise RuntimeError("use_cuda requested but CUDA is not available")
        spec = importlib.util.spec_from_file_location("pcg_repo_test_module", os.path.join(repo, "test.py"))     # not `import test` (stdlib name clash)
        tmod = importlib.util.module_from_spec(spec)
        spec.loader.exec_module(tmod)                                  # defines main()/sample(); nothing runs
        from utils.DataLoader import DataLoader
        from utils.helper import get_model, position_change_seq, revert_postion_change_seq2
        from utils.Interaction import getSequenceInteractionGridMask
        import argparse as _ap

        def seed_all():
            random.seed(seed)
            np.random.seed(seed)
            torch.manual_seed(seed)
            if cuda:
                torch.cuda.manual_seed_all(seed)

        def sync():
            if cuda:
                torch.cuda.synchronize()

        with open(cf_dst, "rb") as fh:
            saved_args = pickle.load(fh)
        cpu_override = False
        if not cuda and getattr(saved_args, "use_cuda", False):
            saved_args.use_cuda = False             # in-memory only (the model calls .cuda() when this is True); the config FILE is not changed
            cpu_override = True
        sample_args = _ap.Namespace(obs_length=6, pred_length=6, use_cuda=cuda, method=method, epoch=epoch, sample_size=1)
        t = time.perf_counter()
        dataloader = DataLoader(1, sample_args.obs_length + sample_args.pred_length, infer=True, filtering=True)
        dataloader.reset_batch_pointer()
        net = get_model(method, saved_args, True)
        if cuda:
            net = net.cuda()
        try:
            ck = torch.load(ck_dst, map_location=("cuda" if cuda else "cpu"), weights_only=False)
        except TypeError:
            ck = torch.load(ck_dst, map_location=("cuda" if cuda else "cpu"))
        net.load_state_dict(ck["state_dict"])

        def model_state() -> Dict[str, Any]:
            d = [m for m in net.modules() if isinstance(m, torch.nn.Dropout)]
            return {"net_training": bool(net.training), "dropout_modules": len(d), "dropout_modules_in_training_mode": sum(1 for m in d if m.training),
                    "dropout_probabilities": sorted({float(m.p) for m in d})}
        state_at_load = model_state()                                  # what get_model + load_state_dict left (test.py never changes it)
        if eval_mode:
            net.eval()                                                  # only on explicit request; recorded below
        state_before_timing = model_state()
        scope = sample_grad_scope(os.path.join(repo, "test.py"))
        setup_s = time.perf_counter() - t
        t = time.perf_counter()
        whole = dataloader.batch_creater(False, method, suffle=False)
        batch_creation_s = time.perf_counter() - t
        x_W, nP_W, P_W, xv_W, nV_W, V_W = whole[0], whole[1], whole[2], whole[3], whole[4], whole[5]
        n_total = dataloader.num_batches
        n_use = n_total if max_trajectories is None else min(n_total, max_trajectories)
        if n_use <= warmup:
            raise RuntimeError("need more trajectories (%d) than warm-up iterations (%d)" % (n_use, warmup))
        obs = sample_args.obs_length
        timers: Dict[str, float] = {"fwd": 0.0, "calls": 0}
        gobs: Dict[str, Any] = {"forward": set(), "prep": set(), "post": set(), "prep_tensors_require_grad": False, "output_tensors_require_grad": False,
                                "output_has_grad_fn": False}

        def pre_hook(mod, inp):
            gobs["forward"].add((bool(torch.is_grad_enabled()), bool(torch.is_inference_mode_enabled())))   # (grad_enabled, inference_mode) INSIDE the forward call
            sync()
            timers["t"] = time.perf_counter()

        def post_hook(mod, inp, outp):
            sync()
            timers["fwd"] += time.perf_counter() - timers["t"]
            timers["calls"] += 1

        def one(b: int, stage_times: Optional[Dict[str, float]]) -> Dict[str, Any]:
            x, nP, P = x_W[b], nP_W[b], P_W[b]
            xv, nV, V = xv_W[b], nV_W[b], V_W[b]
            x_seq, numPedsList_seq, PedsList_seq = x[0], nP[0], P[0]
            x_seq_veh, numVehsList_seq, VehsList_seq = xv[0], nV[0], V[0]
            gobs["prep"].add((bool(torch.is_grad_enabled()), bool(torch.is_inference_mode_enabled())))
            sync()
            t0 = time.perf_counter()
            # --- prep (replicates test.py: dense vector creation, grid masks on CPU, position change)
            x_seq, lookup_seq = dataloader.convert_proper_array(x_seq, numPedsList_seq, PedsList_seq)
            x_seq_veh, lookup_seq_veh = dataloader.convert_proper_array(x_seq_veh, numVehsList_seq, VehsList_seq, veh_flag=True)
            orig_x_seq = x_seq.clone()
            grid_seq, grid_TTC_seq = getSequenceInteractionGridMask(x_seq, PedsList_seq, x_seq, PedsList_seq, saved_args.TTC, saved_args.D_min,
                                                                    saved_args.num_sector, False, lookup_seq, lookup_seq)
            grid_seq_veh_in_ped, grid_TTC_veh_seq = getSequenceInteractionGridMask(x_seq, PedsList_seq, x_seq_veh, VehsList_seq, saved_args.TTC_veh,
                                                                                    saved_args.D_min_veh, saved_args.num_sector, False, lookup_seq,
                                                                                    lookup_seq_veh, is_heterogeneous=True, is_occupancy=False)
            x_seq, first_values_dict = position_change_seq(x_seq, PedsList_seq, lookup_seq)
            gobs["prep_tensors_require_grad"] = gobs["prep_tensors_require_grad"] or bool(x_seq.requires_grad or x_seq_veh.requires_grad or orig_x_seq.requires_grad)
            t1 = time.perf_counter()
            # --- transfer
            if cuda:
                x_seq = x_seq.cuda()
                x_seq_veh = x_seq_veh.cuda()
            sync()
            t2 = time.perf_counter()
            # --- rollout: the unchanged baseline sample() with ONE sample
            ret_x_seq, dist_param_seq = tmod.sample(x_seq[:obs], PedsList_seq[:obs], sample_args, net, x_seq, PedsList_seq, saved_args, dataloader,
                                                    lookup_seq, numPedsList_seq, saved_args.gru, first_values_dict, orig_x_seq, grid_seq[:obs], x_seq_veh,
                                                    VehsList_seq, lookup_seq_veh, grid_seq_veh_in_ped[:obs], grid_TTC_seq[:obs], grid_TTC_veh_seq[:obs])
            sync()
            t3 = time.perf_counter()
            gobs["output_tensors_require_grad"] = gobs["output_tensors_require_grad"] or bool(ret_x_seq.requires_grad or dist_param_seq.requires_grad)
            gobs["output_has_grad_fn"] = gobs["output_has_grad_fn"] or (ret_x_seq.grad_fn is not None or dist_param_seq.grad_fn is not None)
            gobs["post"].add((bool(torch.is_grad_enabled()), bool(torch.is_inference_mode_enabled())))
            # --- post (revert position change)
            ret_x_seq = revert_postion_change_seq2(ret_x_seq.cpu(), PedsList_seq, lookup_seq, first_values_dict, orig_x_seq, obs, infer=True)
            dist_param_seq[:, :, 0:2] = revert_postion_change_seq2(dist_param_seq[:, :, 0:2].cpu(), PedsList_seq, lookup_seq, first_values_dict, orig_x_seq,
                                                                   obs, infer=True)
            sync()
            t4 = time.perf_counter()
            finite = bool(torch.isfinite(ret_x_seq).all().item())
            return {"n_peds": len(lookup_seq), "n_vehicles": len(lookup_seq_veh), "prep_s": t1 - t0, "transfer_s": t2 - t1, "rollout_s": t3 - t2,
                    "post_s": t4 - t3, "pipeline_s": t4 - t0, "output_finite": finite, "output_shape": list(ret_x_seq.shape)}

        grad_before, inf_before = torch.is_grad_enabled(), torch.is_inference_mode_enabled()
        # ---------------- pass 1: pipeline latency
        seed_all()
        for b in range(warmup):
            one(b, None)
        sync()
        if cuda:
            torch.cuda.reset_peak_memory_stats()
        rows: List[Dict[str, Any]] = []
        sync()
        loop_t0 = time.perf_counter()
        for b in range(warmup, n_use):
            r = one(b, None)
            r["trajectory_index"] = b
            rows.append(r)
        sync()
        loop_wall = time.perf_counter() - loop_t0
        peak_alloc = torch.cuda.max_memory_allocated() / 2 ** 20 if cuda else None
        peak_res = torch.cuda.max_memory_reserved() / 2 ** 20 if cuda else None
        # ---------------- pass 2: model-only (forward hooks, same trajectories, same seed)
        h1, h2 = net.register_forward_pre_hook(pre_hook), net.register_forward_hook(post_hook)
        try:
            seed_all()
            for b in range(warmup):
                one(b, None)
            fwd = []
            calls = []
            for i, b in enumerate(range(warmup, n_use)):
                timers["fwd"], timers["calls"] = 0.0, 0
                one(b, None)
                rows[i]["model_forward_s"], rows[i]["n_forward_calls"] = timers["fwd"], timers["calls"]
                fwd.append(timers["fwd"])
                calls.append(timers["calls"])
        finally:
            h1.remove()
            h2.remove()
        state_after = model_state()
        if state_after != state_before_timing:
            raise RuntimeError("the benchmark changed the model's train/eval state: before %s, after %s" % (state_before_timing, state_after))
        params = list(net.parameters())
        fwd_modes = sorted(gobs["forward"])
        grad_in_forward = {g for g, _ in gobs["forward"]}
        if not gobs["forward"]:
            fwd_conclusion = "no model forward was observed"
        elif grad_in_forward == {False}:
            fwd_conclusion = "gradient tracking DISABLED during every observed model forward (by the baseline sample() torch.no_grad())"
        elif grad_in_forward == {True}:
            fwd_conclusion = "gradient tracking was ENABLED during every observed model forward"
        else:
            fwd_conclusion = "gradient tracking was enabled in some model forwards and disabled in others"
        gradient_tracking = {
            "benchmark_added_context": "none (no extra torch.no_grad / torch.inference_mode; identical to test.py)",
            "torch_grad_enabled_before_benchmark": bool(grad_before), "torch_inference_mode_enabled_before_benchmark": bool(inf_before),
            "baseline_sample_function_static_check": scope,
            "observed_inside_model_forward": [{"grad_enabled": g, "inference_mode": i} for g, i in fwd_modes],
            "observed_at_prep_stage_outside_sample": [{"grad_enabled": g, "inference_mode": i} for g, i in sorted(gobs["prep"])],
            "observed_at_post_stage_outside_sample": [{"grad_enabled": g, "inference_mode": i} for g, i in sorted(gobs["post"])],
            "prep_tensors_require_grad": gobs["prep_tensors_require_grad"], "output_tensors_require_grad": gobs["output_tensors_require_grad"],
            "output_has_grad_fn": gobs["output_has_grad_fn"],
            "parameters": {"count": len(params), "requiring_grad": sum(1 for q in params if q.requires_grad), "with_gradient_populated": sum(1 for q in params if q.grad is not None)},
            "conclusion": fwd_conclusion + "; the stages outside sample() (preparation, post-processing) ran with grad mode "
                          + ("enabled" if any(g for g, _ in gobs["prep"] | gobs["post"]) else "disabled")
                          + " and created autograd graph: %s; parameters keep requires_grad=True but no gradient was computed or stored"
                          % ("none (no tensor required grad)" if not (gobs["prep_tensors_require_grad"] or gobs["output_tensors_require_grad"] or gobs["output_has_grad_fn"])
                             else "YES - inspect the observation fields")}
        train_eval_state = {"at_load": state_at_load, "before_timing": state_before_timing, "after_all_passes": state_after,
                            "unchanged_during_benchmark": state_after == state_before_timing, "eval_called_by_benchmark": bool(eval_mode),
                            "changed_by_benchmark_relative_to_load": state_before_timing != state_at_load}
        fields = ["trajectory_index", "n_peds", "n_vehicles", "prep_s", "transfer_s", "rollout_s", "post_s", "pipeline_s", "model_forward_s", "n_forward_calls",
                  "output_finite"]
        with open(os.path.join(out, "inference_latency_samples.csv"), "x", newline="", encoding="utf-8") as fh:
            w = csv.DictWriter(fh, fieldnames=fields)
            w.writeheader()
            for r in rows:
                w.writerow({k: r.get(k) for k in fields})
        n = len(rows)
        summary = {
            "benchmark_version": BENCH_VERSION, "run_id": run_id, "finished_utc": _utc(), "scope": SCOPE,
            "environment": {"python_executable": sys.executable, "python_version": sys.version, "platform": platform.platform(), "torch": torch.__version__,
                            "torch_cuda": torch.version.cuda, "numpy": np.__version__, "cuda": cuda,
                            "gpu_name": torch.cuda.get_device_name(0) if cuda else None,
                            "gpu_total_vram_mib": torch.cuda.get_device_properties(0).total_memory / 2 ** 20 if cuda else None,
                            "gpu_driver": _gpu_driver(), "hardware_label": hardware_label,
                            "cudnn_deterministic": torch.backends.cudnn.deterministic, "cudnn_benchmark": torch.backends.cudnn.benchmark},
            "protocol": {"epoch": epoch, "method": method, "sample_size": 1, "obs_length": obs, "pred_length": sample_args.pred_length, "seed": seed,
                         "seed_applies_to": "python random, numpy, torch, torch.cuda (set before each pass; not bitwise-deterministic across hardware)",
                         "warmup_trajectories_excluded": warmup, "timed_trajectories": n, "trajectories_available": n_total, "eval_mode": eval_mode,
                         "net_eval_called": eval_mode, "dropout_active_in_prediction_path": state_before_timing["dropout_modules_in_training_mode"] > 0 and state_before_timing["dropout_probabilities"] != [0.0],
                         "dropout_probability": getattr(saved_args, "dropout", None), "ground_truth_scoring_included": False, "oracle_selection_included": False,
                         "batch_size_of_input": 1, "saved_args_use_cuda_overridden_to_false_in_memory": cpu_override},
            "gradient_tracking": gradient_tracking, "model_train_eval_state": train_eval_state,
            "hashes": hashes, "setup_s": setup_s, "test_set_batch_creation_s": batch_creation_s,
            "latency_s": {"pipeline": describe([r["pipeline_s"] for r in rows]), "prep": describe([r["prep_s"] for r in rows]),
                          "transfer": describe([r["transfer_s"] for r in rows]), "rollout_sample_call": describe([r["rollout_s"] for r in rows]),
                          "post": describe([r["post_s"] for r in rows]), "model_forward_only_sum_per_trajectory": describe(fwd),
                          "model_forward_calls_per_trajectory": {"min": min(calls), "max": max(calls), "mean": sum(calls) / len(calls)}},
            "throughput_sequences_per_s": {"pipeline_pass": (n / loop_wall) if loop_wall > 0 else None, "definition": "timed trajectories / wall time of the timed loop (pass 1)"},
            "peak_gpu_memory_mib": {"max_allocated": peak_alloc, "max_reserved": peak_res, "note": "pass 1 after warm-up; None on CPU"},
            "output_check": {"non_finite_outputs": sum(1 for r in rows if not r["output_finite"]), "shapes_seen": sorted({tuple(r["output_shape"]) for r in rows})[:5]},
            "limitations": ["batch size 1, as in test.py; no batching study", "future actor information is still consumed by the baseline path",
                            "model-only pass uses forward hooks that synchronise, so it is separate from the pipeline pass",
                            "one dataset (HBS test split, 1081 sequences) and one device; no claim beyond this",
                            "no torch.inference_mode variant was measured: the baseline already runs sample() under torch.no_grad(), so an extra context would change nothing that is timed except the preparation/post stages"]}
        with open(os.path.join(out, "inference_benchmark_summary.json"), "x", encoding="utf-8") as fh:
            json.dump(summary, fh, indent=2, sort_keys=True, default=str)
        return summary
    finally:
        os.chdir(old_cwd)
        if sys.path and sys.path[0] == repo:
            sys.path.pop(0)


def main(argv: Optional[Sequence[str]] = None) -> int:
    ap = argparse.ArgumentParser(description="Single-sample PCG predictor benchmark (isolated; never modifies the repository or checkpoints).")
    ap.add_argument("--repo", required=True)
    ap.add_argument("--workdir", required=True)
    ap.add_argument("--run-id", required=True)
    ap.add_argument("--checkpoint", required=True)
    ap.add_argument("--config", required=True)
    ap.add_argument("--epoch", type=int, required=True)
    ap.add_argument("--warmup", type=int, default=20)
    ap.add_argument("--max-trajectories", type=int, default=None)
    ap.add_argument("--seed", type=int, default=12345)
    ap.add_argument("--cpu", action="store_true")
    ap.add_argument("--eval-mode", action="store_true")
    ap.add_argument("--hardware-label", default=None)
    ap.add_argument("--expected-checkpoint-sha256", default=None)
    ap.add_argument("--expected-config-sha256", default=None)
    a = ap.parse_args(argv)
    try:
        s = run_benchmark(a.repo, a.workdir, a.run_id, a.checkpoint, a.config, epoch=a.epoch, warmup=a.warmup, max_trajectories=a.max_trajectories, seed=a.seed,
                          use_cuda=False if a.cpu else None, eval_mode=a.eval_mode, expected_checkpoint_sha256=a.expected_checkpoint_sha256,
                          expected_config_sha256=a.expected_config_sha256, hardware_label=a.hardware_label)
    except RuntimeError as exc:
        print(str(exc), file=sys.stderr)
        return 2
    print(json.dumps({"timed": s["protocol"]["timed_trajectories"], "pipeline_p50_s": s["latency_s"]["pipeline"].get("p50_s")}, indent=2))
    return 0


if __name__ == "__main__":
    sys.exit(main())
''',
}

HELPER_SHA256 = {}
for _name, _src in HELPER_SOURCES.items():
    _p = os.path.join(TOOLS_DIR, _name)
    with open(_p, "x", encoding="utf-8", newline="\n") as _f:
        _f.write(_src)
    HELPER_SHA256[_name] = hashlib.sha256(_src.encode("utf-8")).hexdigest()
sys.path.insert(0, TOOLS_DIR)
import measurement_lib as ml
import monitored_train_runner as runner_mod
import inference_benchmark as bench_mod

print("Measurement run id :", MEAS_RUN_ID)
print("Output directory   :", MEAS_DIR)
print("Kernel interpreter :", sys.executable, "| Python", platform.python_version())
print("Repository present :", os.path.isdir(REPO), "| Drive backup present:", os.path.isdir(DRIVE_BACKUP))
print("Helper SHA-256     :", json.dumps(HELPER_SHA256, indent=1))
print("\nSwitches: RUN_MONITORED_TRAINING =", RUN_MONITORED_TRAINING, "| RUN_INFERENCE_BENCHMARK =", RUN_INFERENCE_BENCHMARK)
```

```python
# ==============================================================================================
# PART B - EXPERIMENT FINGERPRINT  ->  experiment_environment.json   (read-only on the experiment)
# ==============================================================================================
import re, glob, random

def _run(cmd, timeout=15):
    try:
        r = subprocess.run(cmd, capture_output=True, text=True, timeout=timeout)
        return r.stdout.strip() if r.returncode == 0 else "ERROR(%d): %s" % (r.returncode, (r.stderr or r.stdout).strip()[:300])
    except Exception as exc:
        return "UNAVAILABLE: %s" % exc

def _version(mod):
    try:
        m = __import__(mod)
        return getattr(m, "__version__", "unknown")
    except Exception as exc:
        return "not importable: %s" % type(exc).__name__

def _probe_other_interpreter(py):
    """Ask ANOTHER interpreter (e.g. the pcg-env venv) what it really is. Read-only."""
    if not py or not os.path.isfile(py):
        return {"path": py, "exists": False}
    code = ("import sys,json,platform\nd={'executable':sys.executable,'python':sys.version,'platform':platform.platform()}\n"
            "for m in ('torch','torchvision','numpy','pandas','scipy','sklearn','matplotlib','psutil'):\n"
            "    try:\n        mod=__import__(m); d[m]=getattr(mod,'__version__','unknown')\n    except Exception as e:\n        d[m]='not importable: '+type(e).__name__\n"
            "try:\n    import torch; d['cuda_available']=torch.cuda.is_available(); d['torch_cuda']=torch.version.cuda\n"
            "except Exception: pass\nprint(json.dumps(d))")
    out = _run([py, "-c", code], timeout=120)
    try:
        d = json.loads(out.splitlines()[-1]); d["path"] = py; d["exists"] = True
        return d
    except Exception:
        return {"path": py, "exists": True, "probe_output": out[:500]}

def _cfg_dict(path):
    if not os.path.isfile(path):
        return {"path": path, "exists": False}
    with open(path, "rb") as fh:
        ns = pickle.load(fh)                      # a Namespace written by train.py (our own artifact)
    return {"path": path, "exists": True, "sha256": ml.sha256_file(path), "args": {k: (v if isinstance(v, (int, float, str, bool, list, type(None))) else str(v)) for k, v in vars(ns).items()}}

def collect_fingerprint():
    fp = {"fingerprint_version": "pcg-experiment-fingerprint-1.0", "measurement_run_id": MEAS_RUN_ID, "collected_utc": ml.utc_now_iso()}
    # --- interpreters (the historical notebook used two)
    fp["kernel_interpreter"] = {"executable": sys.executable, "python": sys.version, "implementation": platform.python_implementation()}
    fp["pcg_env_interpreter"] = _probe_other_interpreter("/content/pcg-env/bin/python")
    fp["interpreter_for_new_runs"] = PYTHON_FOR_NEW_RUNS or sys.executable
    # --- OS / hardware
    osr = {}
    try:
        with open("/etc/os-release") as fh:
            osr = dict(l.strip().split("=", 1) for l in fh if "=" in l)
    except Exception:
        pass
    fp["os"] = {"platform": platform.platform(), "system": platform.system(), "release": platform.release(), "os_release": {k: v.strip('"') for k, v in osr.items() if k in ("PRETTY_NAME", "VERSION_ID")}}
    cpu = _run(["lscpu"])
    fp["cpu"] = {"logical_cores": os.cpu_count(), "lscpu_model_name": next((l.split(":", 1)[1].strip() for l in cpu.splitlines() if l.startswith("Model name")), None), "platform_processor": platform.processor()}
    try:
        import psutil
        vm = psutil.virtual_memory()
        fp["ram"] = {"total_gib": round(vm.total / 2 ** 30, 2), "available_gib_at_collection": round(vm.available / 2 ** 30, 2)}
    except Exception as exc:
        fp["ram"] = {"unavailable": str(exc)}
    # --- libraries as seen by the KERNEL interpreter
    libs = {m: _version(m) for m in ("torch", "torchvision", "numpy", "pandas", "scipy", "sklearn", "matplotlib", "psutil")}
    fp["kernel_libraries"] = libs
    try:
        import torch
        fp["cuda"] = {"cuda_available": torch.cuda.is_available(), "torch_cuda_version": torch.version.cuda, "cudnn_version": (torch.backends.cudnn.version() if torch.cuda.is_available() else None)}
        if torch.cuda.is_available():
            p = torch.cuda.get_device_properties(0)
            fp["gpu"] = {"name": p.name, "total_vram_mib": p.total_memory // 2 ** 20, "compute_capability": "%d.%d" % (p.major, p.minor), "multiprocessors": p.multi_processor_count}
        fp["determinism_settings_in_kernel"] = {"cudnn_deterministic": torch.backends.cudnn.deterministic, "cudnn_benchmark": torch.backends.cudnn.benchmark,
                                                  "deterministic_algorithms_enabled": torch.are_deterministic_algorithms_enabled()}
    except Exception as exc:
        fp["cuda"] = {"error": str(exc)}
    smi = _run(["nvidia-smi", "--query-gpu=name,memory.total,driver_version", "--format=csv,noheader"])
    fp["nvidia_smi_query"] = smi
    fp["gpu_driver_version"] = smi.split(",")[-1].strip() if smi and not smi.startswith(("ERROR", "UNAVAILABLE")) else None
    # --- git
    fp["git"] = {"commit": _run(["git", "-C", REPO, "rev-parse", "HEAD"]), "branch": _run(["git", "-C", REPO, "rev-parse", "--abbrev-ref", "HEAD"]),
                 "status_porcelain": _run(["git", "-C", REPO, "status", "--porcelain"]).splitlines()}
    fp["git"]["working_tree_clean"] = (len(fp["git"]["status_porcelain"]) == 0)
    # --- training configuration (the NEW run's config and the ORIGINAL CollisionGrid config)
    fp["training_configurations"] = {"training_run_config": _cfg_dict(os.path.join(STORE, "model", "config.pkl")),
                                     "collisiongrid_config_in_repo": _cfg_dict(os.path.join(STORE, "model", "CollisionGrid", "config.pkl")),
                                     "drive_backup_config": _cfg_dict(os.path.join(DRIVE_BACKUP, "config.pkl")),
                                     "drive_backup_original_collisiongrid_config": _cfg_dict(os.path.join(DRIVE_BACKUP, "CollisionGrid_original_config.pkl"))}
    # --- datasets, splits
    data_files = ["Data/HBS/hbs.csv", "Data/train/trajectories.cpkl", "Data/train/trajectories_train.cpkl", "Data/test/trajectories_test.cpkl"]
    fp["datasets"] = [ml.file_record(os.path.join(REPO, f), label=f) for f in data_files]
    fp["split_files"] = {"train_split": "Data/train/trajectories_train.cpkl", "test_split": "Data/test/trajectories_test.cpkl",
                         "note": "the historical loader has no separate validation split (see Phase 2 audit)"}
    # --- source files that define the experiment
    src_files = ["train.py", "test.py", "utils/DataLoader.py", "utils/helper.py", "utils/Interaction.py", "utils/grid.py", "model/model_CollisionGrid.py", "visualization.py", "requirements.txt"]
    fp["source_files"] = [ml.file_record(os.path.join(REPO, f), label=f) for f in src_files]
    # --- seeds: look for seeding in the code instead of assuming
    hits = []
    for f in glob.glob(os.path.join(REPO, "**", "*.py"), recursive=True):
        try:
            for n, line in enumerate(open(f, encoding="utf-8", errors="replace"), 1):
                if re.search(r"manual_seed|np\.random\.seed|random\.seed|set_seed|cudnn\.deterministic|use_deterministic_algorithms", line):
                    hits.append("%s:%d: %s" % (os.path.relpath(f, REPO), n, line.strip()[:120]))
        except OSError:
            pass
    fp["seeds"] = {"seed_calls_found_in_repository_python_files": hits, "conclusion": ("no seeding or determinism call found: the historical training and test were NOT seeded"
                   if not hits else "seeding calls exist; inspect them"), "PYTHONHASHSEED": os.environ.get("PYTHONHASHSEED"), "seed_used_by_this_notebook": None,
                   "note": "new instrumented runs record any seed they set; the monitored training runner sets none (it runs unmodified train.py)"}
    fp["helpers_sha256"] = HELPER_SHA256
    return fp

FINGERPRINT = collect_fingerprint()
_sha = ml.write_new_json(os.path.join(MEAS_DIR, "experiment_environment.json"), FINGERPRINT)
print("wrote experiment_environment.json  sha256 =", _sha)
print("kernel   :", FINGERPRINT["kernel_interpreter"]["executable"], "| torch", FINGERPRINT["kernel_libraries"]["torch"])
print("pcg-env  :", {k: FINGERPRINT["pcg_env_interpreter"].get(k) for k in ("exists", "executable", "torch", "torch_cuda")})
print("git      :", FINGERPRINT["git"]["commit"], "| clean:", FINGERPRINT["git"]["working_tree_clean"])
print("seeds    :", FINGERPRINT["seeds"]["conclusion"])
```
